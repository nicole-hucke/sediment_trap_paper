```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
def plot_distribution(df, title=None, ax=None):
    same = df['same'].dropna()
    different = df['different'].dropna()

    bins = np.linspace(0, 1, 11)
    if ax is None:
        fig, ax = plt.subplots(figsize=(6, 4))

    sns.histplot(same, bins=bins, stat='probability', alpha=0.5, label='Same trap type', kde=True, ax=ax)
    sns.histplot(different, bins=bins, stat='probability', alpha=0.5, label='Different trap type', kde=True, ax=ax)
    ax.set_xlabel('Percent Difference')
    ax.set_ylabel('Proportion')
    if title:
        ax.set_title(title)
    ax.legend()
    return ax
```

## Total Sediment

Summer

Spring

```python
df_spring = pd.read_csv('total_spring.csv')
plot_distribution(df_spring, title='Total Sediment - Spring')
plt.tight_layout()
plt.show()
```

```python
df_summer = pd.read_csv('coarsesand_summer.csv')
plot_distribution(df_summer, title='Coarse Sand - Summer')
plt.tight_layout()
plt.show()
```
