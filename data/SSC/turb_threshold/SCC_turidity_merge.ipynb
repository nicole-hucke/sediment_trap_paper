```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import os

sp_sample_SSC = pd.read_csv('../original_SSC/samples_spring_SSC_full.csv', index_col=1, parse_dates=True)
sm_sample_SSC = pd.read_csv('../original_SSC/samples_summer_SSC.csv', index_col=1, parse_dates=True)
sp_turbidity = pd.read_csv('../turb_SP23_clean_complete.csv', index_col=0, parse_dates=True)
sm_turbidity = pd.read_csv('../turb_SM23_clean_complete.csv', index_col=0, parse_dates=True)
```

```python
sm_turbidity = sm_turbidity[~sm_turbidity.index.duplicated(keep='first')]
# resample turbidity data to 1 minute intervals and take the mean
sp_turbidity_resampled = sp_turbidity.resample('1min').interpolate()
sm_turbidity_resampled = sm_turbidity.resample('1min').interpolate()
sp_turbidity_resampled
```

```python
# join the turbidity data with the SSC samples based on the timestamp
sp_merged = pd.merge(sp_sample_SSC, sp_turbidity_resampled, left_index=True, right_index=True, how='inner')
sm_merged = pd.merge(sm_sample_SSC, sm_turbidity_resampled, left_index=True, right_index=True, how='inner')
```

```python
# export the merged data to csv
sp_merged.to_csv('sp_merged_SSC_turbidity.csv')
sm_merged.to_csv('sm_merged_SSC_turbidity.csv')
```
